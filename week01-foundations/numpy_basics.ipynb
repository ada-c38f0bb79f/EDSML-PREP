## 2. Boolean masks and missing-value sentinels

A comparison creates a Boolean mask. The mask can then select matching elements.

```python
rainfall = np.array([2.5, -999, 0.0, 4.5, -999])
valid_mask = rainfall != -999
valid_rainfall = rainfall[valid_mask]

print(valid_mask)
print(valid_rainfall)
print(valid_rainfall.mean())
```

Do not automatically reject every negative environmental measurement. Negative temperatures, elevations, and anomalies may be valid. Exclude the documented sentinel value precisely.

## 3. Two-dimensional arrays, shapes, and axes

Each row below represents a day. Each column represents a weather station.

```python
temperatures = np.array([
    [18, 20, 22],
    [19, 21, 24],
    [17, 23, 25],
])

print(temperatures.shape)
print("Station means:", temperatures.mean(axis=0))
print("Daily means:", temperatures.mean(axis=1))
```

- `axis=0` removes the row dimension and produces one result per column.
- `axis=1` removes the column dimension and produces one result per row.

```python
daily_highs = temperatures.max(axis=1)
station_lows = temperatures.min(axis=0)

print("Daily highs:", daily_highs)
print("Station lows:", station_lows)
```

## 4. Broadcasting and new dimensions

To compare each row with its own mean, turn the one-dimensional array of daily means into a column with `[:, None]`.

```python
daily_means = temperatures.mean(axis=1)
daily_means_column = daily_means[:, None]

print("Original shape:", daily_means.shape)
print("Column shape:", daily_means_column.shape)
print(daily_means_column)

above_daily_mean_mask = temperatures > daily_means_column
print(above_daily_mean_mask)
print(temperatures[above_daily_mean_mask])
```

`values[:, None]` creates a column with shape `(n, 1)`. `values[None, :]` creates a row with shape `(1, n)`. Check shapes before broadcasting, especially when working with square arrays.

## 5. Two-dimensional slicing

NumPy uses `array[rows, columns]`. Python indices start at zero, and the end of a slice is excluded.

```python
print("Second station:", temperatures[:, 1])
print("Second day:", temperatures[1, :])
print("Top-right block:\n", temperatures[:2, 1:])
print("Bottom-right block:\n", temperatures[1:, 1:])
```

## 6. NaN-aware statistics

`NaN` is a standard floating-point representation of missing data. Use NumPy's `nan`-aware statistical functions when the array's row and column structure must be preserved.

```python
rainfall_2d = np.array([
    [2.0, np.nan, 5.0],
    [0.0, 3.0, np.nan],
    [4.0, 6.0, 8.0],
])

print("NaN mask:\n", np.isnan(rainfall_2d))
print("Column means:", np.nanmean(rainfall_2d, axis=0))
print("Row maxima:", np.nanmax(rainfall_2d, axis=1))
```

Boolean indexing such as `rainfall_2d[~np.isnan(rainfall_2d)]` returns the selected values as a one-dimensional array. This loses the original row and column structure.
